```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Python Environment Successfully Configured!")
print("=" * 40)
print("Python Version:", sys.version)
print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)
print("Environment Ready!")
```

```python
from google.colab import files


df = pd.read_csv("student-mat.csv", sep=";")

print("Dataset Loaded Successfully!")
display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="G3", bins=15, kde=True)

plt.title("Distribution of Student Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
studytime_avg = df.groupby("studytime")["G3"].mean()

plt.figure(figsize=(8, 5))

sns.barplot(x=studytime_avg.index, y=studytime_avg.values)

plt.title("Study Time vs Average Final Grade")
plt.xlabel("Study Time (1 = Low, 4 = High)")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
print("KEY FINDINGS")
print("=" * 40)

print("Total Students:", len(df))
print("Average Final Grade:", round(df["G3"].mean(), 2))
print("Highest Grade:", df["G3"].max())
print("Lowest Grade:", df["G3"].min())

print("\nAverage Grade by Study Time:")
display(studytime_avg)

print("\nConclusion:")
print("The dataset was successfully loaded and analyzed using Python.")
print("Visualizations helped understand student grade distribution.")
print("Study time can be compared with average student performance.")
```
