```python
# Categorize students based on final grades

def categorize_grade(grade):
    if grade >= 15:
        return "Excellent"
    elif grade >= 10:
        return "Good"
    elif grade >= 5:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["G3"].apply(categorize_grade)

performance_counts = df["Performance"].value_counts()

# Create donut chart
plt.figure(figsize=(8, 8))

plt.pie(
    performance_counts,
    labels=performance_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.45, "edgecolor": "white"},
    pctdistance=0.75
)

plt.title("Student Performance Distribution", fontsize=16, fontweight="bold")

plt.text(
    0, 0,
    f"{len(df)}\nStudents",
    ha="center",
    va="center",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

df = pd.read_csv("student-mat.csv", sep=";")

print("Dataset loaded successfully!")
print("Total Students:", len(df))

display(df.head())
```

```python
# Calculate student performance

average_grade = df["G3"].mean()
highest_grade = df["G3"].max()
lowest_grade = df["G3"].min()

print("Student Performance Analysis")
print("=" * 30)

print("Average Final Grade:", round(average_grade, 2))
print("Highest Final Grade:", highest_grade)
print("Lowest Final Grade:", lowest_grade)

print("\nTotal Students:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
# Categorize student performance

average_grade = df["G3"].mean()

if average_grade >= 15:
    print("Overall Performance: Excellent")
elif average_grade >= 10:
    print("Overall Performance: Good")
else:
    print("Overall Performance: Needs Improvement")
```

```python
# Display the first five students' grades

print("First Five Student Grades:")

for grade in df["G3"].head(5):
    print("Final Grade:", grade)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="G3", bins=15, kde=True)

plt.title("Student Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_average = df.groupby("sex")["G3"].mean()

plt.figure(figsize=(7, 5))

sns.barplot(x=gender_average.index, y=gender_average.values)

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.show()
```
