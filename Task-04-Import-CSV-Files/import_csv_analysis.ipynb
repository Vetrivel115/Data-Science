```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
from google.colab import files

df = pd.read_csv("student-mat.csv", sep=";")

print("CSV file imported successfully!")
display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="sex", hue="sex", legend=False)

plt.title("Student Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks([0, 1], ["Female", "Male"])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="absences", y="G3", hue="sex")

plt.title("Student Absences vs Final Grades")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

grade_data = df[["G1", "G2", "G3"]].melt(
    var_name="Grade Period",
    value_name="Grade"
)

sns.boxplot(data=grade_data, x="Grade Period", y="Grade")

plt.title("Grade Comparison Across Academic Periods")
plt.xlabel("Academic Period")
plt.ylabel("Grade")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="G3", bins=15, kde=True)

plt.title("Student Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
studytime_avg = df.groupby("studytime")["G3"].mean()

plt.figure(figsize=(8, 5))

sns.barplot(x=studytime_avg.index, y=studytime_avg.values)

plt.title("Study Time vs Average Final Grade")
plt.xlabel("Study Time (1 = Low, 4 = High)")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
print("KEY FINDINGS")
print("=" * 40)

print("Total Students:", len(df))
print("Average Final Grade:", round(df["G3"].mean(), 2))
print("Highest Final Grade:", df["G3"].max())
print("Lowest Final Grade:", df["G3"].min())

print("\nAverage Grade by Study Time:")
display(studytime_avg)

print("\nConclusion:")
print("The CSV dataset was successfully imported using Pandas.")
print("Data exploration and statistical analysis were performed.")
print("Visualizations helped identify patterns in student performance.")
```
