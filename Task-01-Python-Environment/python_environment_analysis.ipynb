```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Python Environment Setup Successful!")
print("=" * 50)

print("Python Version:", sys.version)
print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)
print("Matplotlib Version:", plt.matplotlib.__version__)
print("Seaborn Version:", sns.__version__)

print("\nAll required libraries imported successfully!")
```

```python
from google.colab import files


df = pd.read_csv("student-mat.csv", sep=";")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Dataset Information")
print("=" * 50)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Info:")
df.info()
```

```python
print("Missing Values")
print("=" * 50)

print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Records:", df.duplicated().sum())
```

```python
print("Statistical Summary")
print("=" * 50)

df.describe()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="G3", bins=20, kde=True)

plt.title("Student Final Grade Distribution")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
study_grade = df.groupby("studytime")["G3"].mean()

plt.figure(figsize=(8, 5))

sns.barplot(x=study_grade.index, y=study_grade.values)

plt.title("Study Time vs Average Final Grade")
plt.xlabel("Study Time (1 = Low, 4 = High)")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

correlation = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")

plt.show()
```

```python
average_grade = df["G3"].mean()
highest_grade = df["G3"].max()
lowest_grade = df["G3"].min()

average_study_grade = df.groupby("studytime")["G3"].mean()

print("KEY FINDINGS")
print("=" * 50)

print(f"Total Students: {len(df)}")
print(f"Average Final Grade: {average_grade:.2f}")
print(f"Highest Final Grade: {highest_grade}")
print(f"Lowest Final Grade: {lowest_grade}")

print("\nAverage Grade by Study Time:")
print(average_study_grade)

print("\nInterpretation:")
print("1. The dataset contains information about student academic performance.")
print("2. Final grades vary among students.")
print("3. Study time can be compared with students' average final grades.")
print("4. Data visualization helps identify patterns in academic performance.")
```
